```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# %matplotlib inline
import seaborn as sns
```

```python
df=pd.read_csv('/content/IIoT_Smart_Parking_Management.csv',encoding='unicode_escape')
df.shape
```

```python
df['Entry_Time']=df['Entry_Time'].astype('int')
```

```python
df.rename(columns={"Timestamp":"time"},inplace=True)
```

```python
df[['Entry_Time','Exit_Time']].describe()
```

```python
ax=sns.countplot(x='User_Type',data=df)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
parking=df.groupby(['Parking_Spot_ID'],as_index=False)['Parking_Duration'].sum().sort_values(by='Parking_Duration',ascending=False)
sns.set(rc={'figure.figsize':(30,15)})
sns.barplot(x='Parking_Spot_ID',y='Parking_Duration',data=parking)
```

```python
ax=sns.countplot(data=df,x='Parking_Spot_ID',hue='Dynamic_Pricing_Factor')
sns.set(rc={'figure.figsize':(30,15)})
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
parking=df.groupby(['Vehicle_Type_Weight'],as_index=False)['Dynamic_Pricing_Factor'].sum().sort_values(by='Dynamic_Pricing_Factor',ascending=False)
sns.set(rc={'figure.figsize':(30,15)})
sns.barplot(x='Vehicle_Type_Weight',y='Dynamic_Pricing_Factor',data=parking)
```

```python
ax=sns.countplot(x='Vehicle_Type',data=df)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
ax=sns.countplot(x='Electric_Vehicle',data=df)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
parking=df.groupby(['Vehicle_Type'],as_index=False)['Entry_Time'].sum().sort_values(by='Entry_Time',ascending=False).head()
sns.set(rc={'figure.figsize':(30,15)})
sns.barplot(x='Vehicle_Type',y='Entry_Time',data=parking)
```

```python
parking=df.groupby(['Vehicle_Type'],as_index=False)['Exit_Time'].sum().sort_values(by='Exit_Time',ascending=False).head()
sns.set(rc={'figure.figsize':(30,15)})
sns.barplot(x='Vehicle_Type',y='Exit_Time',data=parking)
```

```python
parking=df.groupby(['time'],as_index=False)['Parking_Spot_ID'].sum().sort_values(by='Parking_Spot_ID',ascending=False).head()
sns.set(rc={'figure.figsize':(30,15)})
sns.barplot(x='time',y='Parking_Spot_ID',data=parking)
```
